```python
import pandas as pd
```

```python
df = pd.read_csv('D://tram/dataset (1)/train.csv', nrows = 5, sep=';')
```

```python
def round_time_up_to_hour(time_value):
    hour, minute = map(int, str(time_value).split(':')[:2])
    return hour + (minute > 0)
```

```python
df.head()
```

```python
df['ngpt_route'] = df['ngpt_route'].apply(lambda x: list(x.split())[0])
day = df['tran_date_time'].apply(lambda x: list(x.split())[0])
time = df['tran_date_time'].apply(lambda x: list(x.split())[1])
time = time.apply(round_time_up_to_hour)
df.insert(0, "Day", day)
df.insert(1, "Time", time)
df = df.drop('tran_date_time', axis = 1)
```

```python
df['Day'] = df['Day'].apply(lambda x: x.replace('-','.')) 
```
